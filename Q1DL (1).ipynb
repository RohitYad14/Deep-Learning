```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv")
df.head()
```

```python
X = df[['crim',
        'zn',
        'indus',
        'chas',
        'nox',
        'rm',
        'age',
        'dis',
        'rad',
        'tax',
        'ptratio',
        'b',
        'lstat']].values
Y = df['medv'].values
```

```python
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
model=MLPRegressor(hidden_layer_sizes=(32,16),activation='relu',solver='adam',max_iter=200, alpha=0.001,random_state=42,verbose=0)
model.fit(X_train, Y_train)
Y_pred = model.predict(X_test)
r2 = r2_score(Y_test, Y_pred)
print("R² Score_1:", r2)


model=MLPRegressor(hidden_layer_sizes=(32,16),activation='relu',solver='adam',max_iter=300, alpha=0.0001,random_state=42,verbose=0)
model.fit(X_train, Y_train)
Y_pred = model.predict(X_test)
r2 = r2_score(Y_test, Y_pred)
print("R² Score_2:", r2)

model=MLPRegressor(hidden_layer_sizes=(64,16),activation='relu',solver='adam',max_iter=200, alpha=0.001,random_state=42,verbose=0)
model.fit(X_train, Y_train)
Y_pred = model.predict(X_test)
r2 = r2_score(Y_test, Y_pred)
print("R² Score_3:", r2)

model=MLPRegressor(hidden_layer_sizes=(32,16,8),activation='relu',solver='adam',max_iter=200, alpha=0.001,random_state=42,verbose=0)
model.fit(X_train, Y_train)
Y_pred = model.predict(X_test)
r2 = r2_score(Y_test, Y_pred)
print("R² Score_4:", r2)

model=MLPRegressor(hidden_layer_sizes=(32,),activation='relu',solver='adam',max_iter=200, alpha=0.001,random_state=42,verbose=0)
model.fit(X_train, Y_train)
Y_pred = model.predict(X_test)
r2 = r2_score(Y_test, Y_pred)
print("R² Score_5:", r2)
```
